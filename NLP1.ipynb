**. Cleaning the Text** :text ko lowercase karenge aur unnecessary spaces remove karenge.

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()

    return text
```

**Expanding Contractions**

```python
contractions = {
    "don't": "do not",
    "can't": "cannot",
    "won't": "will not",
    "isn't": "is not",
    "aren't": "are not",
    "wasn't": "was not",
    "weren't": "were not",
    "didn't": "did not",
    "doesn't": "does not",
    "couldn't": "could not",
    "shouldn't": "should not",
    "wouldn't": "would not",
    "I've": "I have",
    "I'll": "I will",
    "I'm": "I am",
    "you're": "you are",
    "they're": "they are"
}

def expand_contractions(text):
    for contraction, expansion in contractions.items():
        text = re.sub(
            r"\b" + re.escape(contraction) + r"\b",
            expansion,
            text,
            flags=re.IGNORECASE
        )

    return text
```

```python
def remove_html(text):
    text = re.sub(r'<.*?>', '', text)
    return text
```

```python
def remove_punctuation(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text
```

**Tokenization :** Breaking sentences into words.

```python
def tokenize_text(text):
    return word_tokenize(text)
```

```python
stop_words = set(stopwords.words('english'))
stop_words.discard('not')
stop_words.discard('no')
stop_words.discard('nor')


def remove_stopwords(tokens):
    return [
        word for word in tokens
        if word not in stop_words
    ]
```

**Lemmatization :** Converting words  into its base form.

```python
lemmatizer = WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]
```

**Pipe_line**

```python
def preprocess_text(text):

    # 1. Lowercase
    text = text.lower()

    # 2. Expand contractions
    text = expand_contractions(text)

    # 3. Remove HTML tags
    text = remove_html(text)

    # 4. Remove unwanted characters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # 5. Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # 6. Tokenization
    tokens = word_tokenize(text)

    # 7. Stop-word removal
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # 8. Lemmatization
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    return tokens
```

**APPLY THE PIPELINE ON THE DATASET**

```python
df['processed_review'] = df['review'].apply(preprocess_text)
```

```python
sentences = df['processed_review'].tolist()
print(type(sentences))
print(type(sentences[0]))
print(sentences[0][:10])
```

**# WORD TO VEC TRAINING**

```python
# WORD TO VEC TRAINING
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4,
    sg=1
)
```
